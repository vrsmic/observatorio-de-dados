```python
import altair as alt
import pandas as pd
```

# Abrindo csv's

```python
df25 = pd.read_csv('cleaned_data.csv') # Dados do Salário Transparente de 2025
df26 = pd.read_csv('raw_data.csv', sep=';') # Dados do Salário Transparente de 2026
```

# Comparando Salário por Ano

```python
# Criando uma nova coluna do ano
df25['Ano'] = '2025'
df26['Ano'] = '2026'

# Concatenando os dois data frames
df = pd.concat([df25, df26])

# Tirando outliers de salário pelos quantis
Q1 = df['salario_base'].quantile(0.25)
Q3 = df['salario_base'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR
df = df[(df['salario_base'] >= limite_inferior) & (df['salario_base'] <= limite_superior)]

# Renomeando coluna de salário
df.rename(columns={'salario_base': 'Salário Base (R$)'}, inplace=True)
df['Salário Base (R$)'] *= 1000
```

```python
# Plotando Gráfico
chart = alt.Chart(df).mark_boxplot(
    extent='min-max',
    size=60
    ).encode(
    x='Ano:N',
    y='Salário Base (R$):Q',
    color='Ano:N'
).properties(
    title='Comparação de Salário por Ano',
    width=1280,
    height=720
).configure_axis(
    labelFontSize=14,  # Tamanho dos números/rótulos no eixo
    titleFontSize=16   # Tamanho do texto do título do eixo
).configure_legend(
    labelFontSize=14,  # Aumenta o tamanho dos textos dos itens da legenda
    titleFontSize=14,  # Aumenta o tamanho do título da legenda
    symbolSize=150     # Aumenta o tamanho dos ícones/símbolos da legenda
).configure_title(
    fontSize=20     # Altera o tamanho do título do gráfico
)

chart.show()
```

# Salário por Tipo de Contrato

```python
# Tirando os tipos de contrato não especificados
df = df[df['tipo_contrato'] != 'Outro']

# Renomeando a coluna tipo de contrato
df.rename(columns={'tipo_contrato': 'Tipo de Contrato'}, inplace=True)
```

```python
# Plotando Gráfico
chart = alt.Chart(df).mark_boxplot(
    extent='min-max',
    size=60
    ).encode(
    x=alt.X('Tipo de Contrato:N', sort=['PJ', 'CLT', 'Estágio']),
    y='Salário Base (R$):Q',
    color='Tipo de Contrato:N'
).properties(
    title='Salário por Tipo de Contrato',
    width=1280,
    height=720
).configure_axis(
    labelFontSize=14,  # Tamanho dos números/rótulos no eixo
    titleFontSize=16   # Tamanho do texto do título do eixo
).configure_legend(
    labelFontSize=14,  # Aumenta o tamanho dos textos dos itens da legenda
    titleFontSize=14,  # Aumenta o tamanho do título da legenda
    symbolSize=150     # Aumenta o tamanho dos ícones/símbolos da legenda
).configure_title(
    fontSize=20     # Altera o tamanho do título do gráfico
)

chart.show()
```

# Salário por Nível

```python
# Renomeando a coluna de nível
df.rename(columns={'nivel': 'Nível'}, inplace=True)
```

```python
# Plotando Gráfico
chart = alt.Chart(df).mark_boxplot(
    extent='min-max',
    size=60
    ).encode(
    x=alt.X('Nível:N', sort=['Líder / Coordenador', 'Especialista', 'Sênior', 'Pleno', 'Júnior', 'Estagiário']),
    y='Salário Base (R$):Q',
    color='Nível:N'
).properties(
    title='Salário por Nível',
    width=1280,
    height=720
).configure_axis(
    labelFontSize=14,  # Tamanho dos números/rótulos no eixo
    titleFontSize=16   # Tamanho do texto do título do eixo
).configure_legend(
    labelFontSize=14,  # Aumenta o tamanho dos textos dos itens da legenda
    titleFontSize=14,  # Aumenta o tamanho do título da legenda
    symbolSize=150     # Aumenta o tamanho dos ícones/símbolos da legenda
).configure_title(
    fontSize=20     # Altera o tamanho do título do gráfico
)

chart.show()
```

```python
# Criando tabela com as médias
media_ordenada = df.groupby('Nível')['Salário Base (R$)'].mean().sort_values(ascending=False).reset_index()

# Formatando os valores
media_ordenada['Salário Base (R$)'] = media_ordenada['Salário Base (R$)'].apply(lambda x: f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))

media_ordenada
```
